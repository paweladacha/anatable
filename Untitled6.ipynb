### implementation of dynamic progrmaming solution to partitioning problem

```python
def partition_dp_exact(items: list[tuple[str, int]]) -> tuple[list[tuple[str, int]], list[tuple[str, int]]]:
    total_weight = sum(w for _, w in items)
    target = total_weight // 2
    n = len(items)

    print(f'Total weight: {total_weight}')
    print(f'Target: {target}')
    print(f'Number of items: {n}')

    # dp[j] przechowywać będzie True/False.
    # Aby odzyskać składniki, przechowujemy ścieżkę (indeks elementu, który pozwolił uzyskać sumę j).
    parent = [-1] * (target + 1)
    dp = [False] * (target + 1)
    dp[0] = True

    print('start')
    print(f'dp: {dp}')
    print(f'parent: {parent}')

    # Wypełnianie tabeli DP (od prawej do lewej, by użyć 1D array)
    for idx, (_, weight) in enumerate(items):
        print(f'idx: {idx}', f'weight: {weight}')
        for j in range(target, weight - 1, -1):
            print(f'j: {j}')
            print(f'at {j - weight}: {dp[j - weight]} and {not dp[j]}')
            if dp[j - weight] and not dp[j]:
                dp[j] = True
                parent[j] = idx
                print(f'dp: {dp}')
                print(f'parent: {parent}')

    print(f'dp: {dp}')
    print(f'parent: {parent}')

    # Szukamy największej osiągalnej sumy <= target
    best_sum = target
    while best_sum > 0 and not dp[best_sum]:
        best_sum -= 1

    print(f'Best sum: {best_sum}')

    # Odzyskiwanie elementów grupy A
    group_a_indices = set()
    curr = best_sum
    while curr > 0:
        idx = parent[curr]
        group_a_indices.add(idx)
        curr -= items[idx][1]
        print(f'curr: {curr}, {items[idx][1]}')

    print(f'Group A indices: {group_a_indices}')

    # Podział na grupę A i B
    group_a = [item for i, item in enumerate(items) if i in group_a_indices]
    group_b = [item for i, item in enumerate(items) if i not in group_a_indices]

    return group_a, group_b
```

```python
def split_lpt(items: list[tuple[str, int]]) -> tuple[list, list]:
    # Jeśli lista nie jest wcześniej posortowana:
    items = sorted(items, key=lambda x: x[1], reverse=True)

    group_a, group_b = [], []
    sum_a = sum_b = 0

    for label,weight in items:
        # item[1] to waga
        if sum_a <= sum_b:
            group_a.append(label)
            sum_a += weight
        else:
            group_b.append(label)
            sum_b += weight

    return (tuple(group_a),sum_a), (tuple(group_b),sum_b)
```

```python
def build_decision_tree(dataframe, ignored_columns=None, max_depth=5, memo=None):
    if ignored_columns is None:
        ignored_columns = ['id', 'description', 'created_date', 'amount', 'retries', 'quantity']
    else:
        ignored_columns = list(ignored_columns)
    if memo is None:
        memo = {}

    # Convert ignored_columns to a tuple for memoization keys
    ignored_key = tuple(sorted(ignored_columns))
    df_key = hash(frozenset(dataframe.index))
    memo_key = (df_key, ignored_key)

    if memo_key in memo:
        return memo[memo_key]

    # 1. Get column counts (and filter out columns to process)
    counts = get_column_counts(dataframe, ignored_columns=ignored_columns)

    # Filter out columns that have 0 or 1 unique values as they can't be split
    valid_counts = {k: v for k, v in counts.items() if len(v) > 1}

    if not valid_counts or max_depth == 0:
        return {"leaf": True, "size": len(dataframe)}

    # Sort columns by unique count
    sorted_counts = sort_by_unique_count(valid_counts)

    # 2. Perform split_lpt for each column to evaluate splits
    splits = {}
    for col, values in sorted_counts:
        splits[col] = split_lpt(values.items())

    # 3. Find the best split (the one with the minimal difference between group weights)
    best_col, (group_a_info, group_b_info) = find_first_best(splits)
    group_a_vals, sum_a = group_a_info
    group_b_vals, sum_b = group_b_info

    # If the chosen column only has 2 unique values, add it to ignored_columns for children
    next_ignored = ignored_columns.copy()
    if len(valid_counts[best_col]) <= 2:
        if best_col not in next_ignored:
            next_ignored.append(best_col)

    # Filter dataframe into two subsets
    # Handling nan values correctly in boolean indexing
    if None in group_a_vals or any(pd.isna(x) for x in group_a_vals):
        mask_a = dataframe[best_col].isin(group_a_vals) | dataframe[best_col].isna()
    else:
        mask_a = dataframe[best_col].isin(group_a_vals) & dataframe[best_col].notna()

    df_a = dataframe[mask_a]
    df_b = dataframe[~mask_a]

    # Recursively build left and right subtrees
    tree = {
        "split_column": best_col,
        "ignored_columns_at_node": ignored_columns,
        "group_a": {
            "values": group_a_vals,
            "weight": sum_a,
            "subtree": build_decision_tree(df_a, next_ignored, max_depth - 1, memo)
        },
        "group_b": {
            "values": group_b_vals,
            "weight": sum_b,
            "subtree": build_decision_tree(df_b, next_ignored, max_depth - 1, memo)
        }
    }

    memo[memo_key] = tree
    return tree

# Run the decision tree on our initial DataFrame
import json
tree_structure = build_decision_tree(df, ignored_columns=['id', 'description', 'created_date', 'amount', 'retries', 'quantity'])
pprint(tree_structure)
```

```python
# get unique values for columns with ignore
def get_column_uniques(dataframe, ignored_columns=None, as_set=True):
    if ignored_columns is None:
        columns_to_process = dataframe.columns
    else:
        columns_to_process = [col for col in dataframe.columns if col not in ignored_columns]

    result = {}

    for col in columns_to_process:
        # Obliczamy unikalne wartości uwzględniając również NaN/None
        unique_vals = dataframe[col].unique()
        if as_set:
            result[col] = set(unique_vals)
        else:
            result[col] = list(unique_vals)

    return result

def get_col_cardinality(uniques_dict):
  return {k:len(v) for k,v in uniques_dict.items()}

# Przykład użycia:
uniques_dict = get_column_uniques(df, ignored_columns=['id', 'description'])
print('\n>>>uniques_dict:')
if len(str(uniques_dict))<1000:
  pprint(uniques_dict)
print('\n>>>cardinality:')
pprint(get_col_cardinality(uniques_dict))
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency

def cramers_v(x, y):
    """Oblicza współczynnik V Craméra dla dwóch zmiennych kategorycznych."""
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    if n == 0:
        return 0.0
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    # Korekta biasu dla małych prób
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    if min((kcorr-1), (rcorr-1)) <= 0:
        return 0.0
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

# Automatyczne rozpoznawanie kolumn kategorycznych
# Wybieramy typy tekstowe/kategoryczne i odrzucamy te o zbyt dużej kardynalności (np. ID, Email)
potential_cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns
cat_cols = [col for col in potential_cat_cols if df[col].nunique() <= 100]

print("Automatycznie wykryte kolumny kategoryczne:", cat_cols)

# Upewniamy się, że braki danych są traktowane jako osobna kategoria
df_cat = df[cat_cols].fillna('Missing').astype(str)

# Obliczenie macierzy powiązań
n_cols = len(cat_cols)
corr_matrix_cat = pd.DataFrame(np.zeros((n_cols, n_cols)), columns=cat_cols, index=cat_cols)

for col1 in cat_cols:
    for col2 in cat_cols:
        if col1 == col2:
            corr_matrix_cat.loc[col1, col2] = 1.0
        else:
            corr_matrix_cat.loc[col1, col2] = cramers_v(df_cat[col1], df_cat[col2])

# Wizualizacja
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix_cat, annot=True, cmap='YlGnBu', fmt='.2f', linewidths=0.5, vmin=0, vmax=1)
plt.title('Asocjacje (V Craméra) między automatycznie wykrytymi zmiennymi kategorycznymi')
plt.tight_layout()
plt.show()
```

### 1. Numeric Representative Tests
**Methodology:**
- **Kolmogorov–Smirnov (KS) Test:** Measures the maximum distance between the cumulative distribution functions (CDFs) of the full and sample datasets. A high p-value ($>0.05$) indicates we cannot reject the null hypothesis that the two samples come from the same distribution.
- <0.03 perfect
- 0.03 - 0.05 - very good
- 0.05 - 0.10 - visible difference
- >0.10 - visible bias
p-value can be ignored here (especially when sample size >100k)
- **Wasserstein Distance (Earth Mover's Distance):** Measures the minimum work needed to transform one distribution into the other. Lower values indicate closer representativeness.
**!this should be normalized wasserstein distance!** to do

```python
from scipy.stats import ks_2samp, wasserstein_distance
import pandas as pd
import numpy as np

def evaluate_numeric_distribution(df_full, df_sample, num_cols):
    results = {}
    for col in num_cols:
        # Drop NaNs for statistical comparison
        full_vals = df_full[col].dropna().values
        sample_vals = df_sample[col].dropna().values

        if len(full_vals) == 0 or len(sample_vals) == 0:
            continue

        ks_stat, p_val = ks_2samp(full_vals, sample_vals)
        wd = wasserstein_distance(full_vals, sample_vals)

        results[col] = {
            'KS Statistic': ks_stat,
            'p-value': p_val,
            'Wasserstein Dist': wd
        }
    return pd.DataFrame(results).T
```

```python
num_cols = df_full.select_dtypes(include=[np.number]).columns.tolist()
# Exclude typical ID columns if applicable
num_cols = [c for c in num_cols if c not in ['Transaction_ID', 'Customer_ID', 'Phone', 'Zipcode']]

numeric_evaluation = evaluate_numeric_distribution(df_full, df_sample, num_cols)
display(numeric_evaluation)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_distribution_comparison(df_full, df_sample, columns, num_cols_per_row=2):
    """
    Plots normalized, overlayed distribution comparisons (using KDE)
    between the full dataset and the sample dataset for specified numeric columns.
    """
    n_cols = len(columns)
    if n_cols == 0:
        print("No columns to plot.")
        return

    n_rows = (n_cols + num_cols_per_row - 1) // num_cols_per_row
    fig, axes = plt.subplots(n_rows, num_cols_per_row, figsize=(15, 4 * n_rows))
    axes = axes.flatten() if n_cols > 1 else [axes]

    for i, col in enumerate(columns):
        ax = axes[i]
        # Plot normalized KDE curves for both full and sample sets with fill=True instead of shade=True
        sns.kdeplot(data=df_full, x=col, label='Full Dataset', color='royalblue', fill=True, alpha=0.4, ax=ax, common_norm=False)
        sns.kdeplot(data=df_sample, x=col, label='Sample Dataset', color='darkorange', fill=True, alpha=0.4, ax=ax, common_norm=False)

        ax.set_title(f'Normalized Distribution: {col}')
        ax.set_xlabel(col)
        ax.set_ylabel('Density')
        ax.legend()
        ax.grid(True, linestyle='--', alpha=0.5)

    # Hide any unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()

# Compare distributions for the specified numeric columns
plot_distribution_comparison(df_full, df_sample, num_cols)
```

### 2. Categorical Representative Tests
**Methodology:**
- **Total Variation Distance (TVD):** $TVD = \frac{1}{2} \sum |P_{\text{full}}(i) - P_{\text{sample}}(i)|$. Values range from 0 (perfect match) to 1.
<0.02 perfect; 0.02 - 0.05 good; >0.10 wrong

- **Population Stability Index (PSI):** Measures shifts in distribution. $PSI < 0.1$ implies minimal shift, $0.1 \le PSI < 0.25$ minor shift, and $PSI \ge 0.25$ major shift.
<0.1 perfect; 0.1-0.25 minor differences; 0.25 different

```python
def calculate_psi(full_probs, sample_probs, epsilon=1e-4):
    # Align indices to cover all classes
    all_classes = full_probs.index.union(sample_probs.index)
    p = full_probs.reindex(all_classes, fill_value=0) + epsilon
    q = sample_probs.reindex(all_classes, fill_value=0) + epsilon
    # Normalize to 1 after epsilon adjustment
    p = p / p.sum()
    q = q / q.sum()
    return np.sum((p - q) * np.log(p / q))

def evaluate_categorical_distribution(df_full, df_sample, cat_cols):
    results = {}
    for col in cat_cols:
        # Include missing values as a distinct category representation
        full_probs = df_full[col].fillna('Missing').value_counts(normalize=True)
        sample_probs = df_sample[col].fillna('Missing').value_counts(normalize=True)

        # Total Variation Distance (TVD)
        all_classes = full_probs.index.union(sample_probs.index)
        p = full_probs.reindex(all_classes, fill_value=0.0)
        q = sample_probs.reindex(all_classes, fill_value=0.0)
        tvd = 0.5 * np.sum(np.abs(p - q))

        # PSI
        psi = calculate_psi(full_probs, sample_probs)

        results[col] = {
            'TVD': tvd,
            'PSI': psi
        }
    return pd.DataFrame(results).T
```

```python
cat_cols = ['Country', 'Customer_Segment', 'Feedback', 'Gender', 'Income', 'Month', 'Order_Status', 'Payment_Method', 'Product_Category', 'Ratings', 'Shipping_Method', 'Year']
categorical_evaluation = evaluate_categorical_distribution(df_full, df_sample, cat_cols)
display(categorical_evaluation)
```

### 3. Edge Cases
- **Rare Category Recall:** Proportion of unique categories found in the sample compared to the full dataset: $\frac{|Unique_{\text{sample}}|}{|Unique_{\text{full}}|}$.
- **Null & Special Value Parity:** Compares the proportions of missing data between the full set and the sample.

```python
def evaluate_edge_cases(df_full, df_sample, columns):
    results = {}
    for col in columns:
        # Unique recall
        unique_full = set(df_full[col].dropna().unique())
        unique_sample = set(df_sample[col].dropna().unique())
        recall = len(unique_sample.intersection(unique_full)) / len(unique_full) if unique_full else 1.0

        # Null parity
        null_ratio_full = df_full[col].isna().mean()
        null_ratio_sample = df_sample[col].isna().mean()

        results[col] = {
            'Rare Category Recall': recall,
            'Null Ratio Full': null_ratio_full,
            'Null Ratio Sample': null_ratio_sample,
            'Null Ratio Diff': abs(null_ratio_full - null_ratio_sample)
        }
    return pd.DataFrame(results).T
```

```python
edge_case_evaluation = evaluate_edge_cases(df_full, df_sample, cat_cols)
display(edge_case_evaluation)
```

```python
import itertools
import pandas as pd
import numpy as np

def evaluate_edge_cases_pairs(df_full, df_sample, columns):
    """
    Evaluates edge cases on all unique pairs of categorical columns,
    measuring joint category recall (percentage of existing value pairs preserved)
    and joint null ratio (percentage of rows where both values are missing).
    """
    pair_results = {}
    # Generate all unique combinations of column pairs
    column_pairs = list(itertools.combinations(columns, 2))

    for col1, col2 in column_pairs:
        # 1. Joint Rare Category Recall (excluding pairs containing NaNs for raw value space recall)
        full_pairs = df_full[[col1, col2]].dropna()
        sample_pairs = df_sample[[col1, col2]].dropna()

        unique_full = set(full_pairs.itertuples(index=False, name=None))
        unique_sample = set(sample_pairs.itertuples(index=False, name=None))

        recall = len(unique_sample.intersection(unique_full)) / len(unique_full) if unique_full else 1.0

        # 2. Joint Null Parity (proportion of rows where both columns are NaN)
        joint_null_full = (df_full[col1].isna() & df_full[col2].isna()).mean()
        joint_null_sample = (df_sample[col1].isna() & df_sample[col2].isna()).mean()

        pair_results[f"{col1} + {col2}"] = {
            'Joint Unique Combinations (Full)': len(unique_full),
            'Joint Unique Combinations (Sample)': len(unique_sample),
            'Joint Rare Pair Recall': recall,
            'Joint Null Ratio Full': joint_null_full,
            'Joint Null Ratio Sample': joint_null_sample,
            'Joint Null Ratio Diff': abs(joint_null_full - joint_null_sample)
        }

    return pd.DataFrame(pair_results).T

# Run the pair-wise edge cases evaluation on our categorical column list
pair_edge_case_evaluation = evaluate_edge_cases_pairs(df_full, df_sample, cat_cols)
pair_edge_case_evaluation_sorted = pair_edge_case_evaluation.sort_values(by='Joint Rare Pair Recall')

print("Top 10 Column Pairs with Lowest Joint Category Recall:")
display(pair_edge_case_evaluation_sorted.head(10))
```

### 4. Classifier Two-Sample Test (C2ST)
We build an adversarial model to detect if the sample can be reliably distinguished from a standard random sample of the same size. If the model's cross-validated AUC remains near 0.50, the structured sample represents the true distribution excellently without introduced bias.

```python
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score

def run_c2st(df_full, df_sample, ignored_columns=None):
    if ignored_columns is None:
        ignored_columns = []

    sample_size = len(df_sample)

    # Generate a random sample of the same size from the full dataset
    df_random = df_full.sample(n=sample_size, random_state=42).copy()

    # Mark datasets
    df_sample_labeled = df_sample.copy()
    df_sample_labeled['target'] = 1

    df_random_labeled = df_random.copy()
    df_random_labeled['target'] = 0

    # Combine
    combined = pd.concat([df_sample_labeled, df_random_labeled], ignore_index=True)

    # Drop ignored/high-cardinality/uninformative features
    features = [c for c in combined.columns if c not in ignored_columns and c != 'target']

    X = combined[features].copy()
    y = combined['target']

    # Object/Category conversions
    for col in X.select_dtypes(include=['object', 'category']).columns:
        X[col] = X[col].astype('category')

    # 5-fold CV
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    oof_preds = np.zeros(len(combined))

    for train_idx, val_idx in skf.split(X, y):
        X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

        clf = lgb.LGBMClassifier(
            n_estimators=100,
            learning_rate=0.05,
            random_state=42,
            verbosity=-1
        )
        clf.fit(
            X_train,
            y_train,
            eval_set=[(X_val, y_val)],
            callbacks=[lgb.early_stopping(stopping_rounds=10, verbose=False)]
        )

        oof_preds[val_idx] = clf.predict_proba(X_val)[:, 1]

    auc_score = roc_auc_score(y, oof_preds)
    print(f"C2ST CV AUC Score: {auc_score:.4f}")
    return auc_score
```

```python
ignored_cols = ['Transaction_ID', 'Customer_ID', 'Name', 'Email', 'Phone', 'Address', 'Zipcode', 'Date', 'Time']
c2st_auc = run_c2st(df_full, df_sample, ignored_columns=ignored_cols)
```
